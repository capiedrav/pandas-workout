# Exercise 16 - Pandemic taxis

In this exercise, I want you to create a data frame from two different CSV files containing New York taxi data: one from July 2019 (before the pandemic) and a second from July 2020 (near the height of the pandemic, at least in New York). The data frame should contain three columns from the files: ```passenger_count```, ```total_amount```, and ```payment_type```. It should also include a fifth column, ```year```, which should be set to 2019 or 2020, depending on the file from which the data was loaded.

* Load data from july 2019:

````python
# NOTE: files nyc_taxy_2019-07.csv and nyc_taxy_2020-7.csv are not included due to its size. Get them from book's repository.

import pandas as pd

taxi_rides_2019 = pd.read_csv(
    "./nyc_taxi_2019-07.csv", # data from july 2019
    header=0,
    usecols=["passenger_count", "total_amount", "payment_type"],
)

taxi_rides_2019["year"] = 2019 # define column year

taxi_rides_2019.head(10)
````

* Load data from july 2020:

````python
taxi_rides_2020 = pd.read_csv(
    "./nyc_taxi_2020-07.csv", # data from july 2020
    header=0,
    usecols=["passenger_count", "total_amount", "payment_type"],
)

taxi_rides_2020["year"] = 2020 # define column year

taxi_rides_2020.head(10)
````

* concat both Dataframes:

````python
taxi_rides = pd.concat([taxi_rides_2019, taxi_rides_2020])

taxi_rides.head(10)
````

With that data in hand, I want you to answer a few questions:

* How many rides were taken in 2019 and 2020, and what is the difference between these two figures?

````python
rides_2019 = taxi_rides.loc[taxi_rides["year"] == 2019, "passenger_count"].count()
rides_2020 = taxi_rides.loc[taxi_rides["year"] == 2020, "passenger_count"].count()

print(f"Rides in 2019: {rides_2019:,.0f}\nRides in 2020: {rides_2020:,.0f}\nDifference: {rides_2019 - rides_2020:,.0f}")
````

* How much money (in total) was collected in 2019 and 2020, and what was the difference between these two figures?

````python
money_2019 = taxi_rides.loc[taxi_rides.year == 2019, "total_amount"].sum()
money_2020 = taxi_rides.loc[taxi_rides.year == 2020, "total_amount"].sum()

print(f"Money 2019: ${money_2019:,.2f}\nMoney 2020: ${money_2020:,.2f}\nDifference: ${money_2019 - money_2020:,.2f}")
````

* Did the proportion of trips with more than one passenger change dramatically?

````python
rides_multiple_passengers_2019 = taxi_rides.query("(passenger_count > 1) & (year == 2019)").shape[0]
rides_multiple_passengers_2019_prop = (rides_multiple_passengers_2019 / rides_2019)

rides_multiple_passengers_2020 = taxi_rides.query("(passenger_count > 1) & (year == 2020)").shape[0]
rides_multiple_passengers_2020_prop = (rides_multiple_passengers_2020 / rides_2020)

print(f"Proportion of more of one passenger rides in 2019: {rides_multiple_passengers_2019_prop:.2f}\n"
      f"Proportion of more of one passenger rides in 2020: {rides_multiple_passengers_2020_prop:.2f}"
      )
````

**Answer:** the proportion of rides with more than one passenger in 2020 decreased in almost a third in comparison with 2019.

* Did people use cash (i.e., ```payment_type``` of 2) less in 2020 than in 2019?

````python
CASH = 2 # cash is payment type number 2
paid_in_cash_2019 = taxi_rides.loc[(taxi_rides.year == 2019) & (taxi_rides["payment_type"] == CASH)].shape[0]
paid_in_cash_2019_prop = paid_in_cash_2019 / rides_2019

paid_in_cash_2020 = taxi_rides.query("(year == 2020) & (payment_type == @CASH)").shape[0]
paid_in_cash_2020_prop = paid_in_cash_2020 / rides_2020

print(f"Proportion of rides paid in cash in 2019: {paid_in_cash_2019_prop:.2f}\n"
      f"Proportion of rides paid in cash in 2020: {paid_in_cash_2020_prop:.2f}\n")
````

**Answer:** the use of cash in 2020 is a little higher than in 2019.
