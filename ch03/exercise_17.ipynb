# Exercise 17 - Setting column types

Once again, I want you to create a data frame based on New York taxi data from January 2020. This time, however, I want to ensure that the data is in the most appropriate and compact form it can be and will use as little memory as possible when being loaded. So, I want you to do the following:

*  Specify the dtype for each column as you read it in.

````python
import pandas as pd
import numpy as np
from numpy.ma.core import shape

taxi_rides = pd.read_csv(
    "./nyc_taxi_2020-01.csv",
    header=0,
    usecols=["passenger_count", "total_amount", "payment_type"],
)

taxi_rides.head(10)
````

*  Identify rows containing ```NaN``` values. Which columns are ```NaN```, and why?

````python
nan_rows = taxi_rides.loc[
    (taxi_rides["passenger_count"].isna()) | \
    (taxi_rides["payment_type"].isna()) | \
    (taxi_rides["total_amount"].isna())
].shape[0]

print(f"Number of rows with NaN values: {nan_rows}\n"
      f"Percentage of rows with NaN values: {(nan_rows / taxi_rides.shape[0]) * 100:.2f}%"
      )
````

* Remove any rows containing any ```NaN``` values.

````python
taxi_rides = taxi_rides.copy().dropna()

nan_rows = taxi_rides.loc[
    (taxi_rides["passenger_count"].isna()) | \
    (taxi_rides["payment_type"].isna()) | \
    (taxi_rides["total_amount"].isna())
    ].shape[0]

nan_rows
````

*  Set the ```dtype``` for each column to the smallest, most appropriate value.

````python
taxi_rides["passenger_count"] = taxi_rides["passenger_count"].astype(np.int8)
taxi_rides["payment_type"] = taxi_rides["payment_type"].astype(np.int8)
taxi_rides["total_amount"] = taxi_rides["total_amount"].astype(np.float32)

taxi_rides.head(10)
````

## Beyond the exercise

* Create a data frame from four other columns (```VendorID```, ```trip_distance```, ```tip_amount```, and ```total_amount```), specifying the ```dtype``` for each. What types are most appropriate? Can you use them directly, or must you first clean the data?

````python
taxi_rides_2 = pd.read_csv(
    "./nyc_taxi_2020-01.csv",
    header=0,
    usecols=["VendorID", "trip_distance", "tip_amount", "total_amount"],
)

taxi_rides_2.head(10)
````

````python
nan_rows = taxi_rides_2.loc[
    taxi_rides_2["VendorID"].isna() | \
    taxi_rides_2["trip_distance"].isna() |
    taxi_rides_2["tip_amount"].isna() | \
    taxi_rides_2["total_amount"].isna()
].shape[0]

print(f"Number of rows with NaN values: {nan_rows:,.0f}\n"
      f"Percentage of of rows with NaN values: {(nan_rows / taxi_rides_2.shape[0]) * 100:.2f}%")
````

There are ```NaN``` values in the dataset, why must clean the data before assigning a ```dtype``` to each column:

````python
taxi_rides_2_clean = taxi_rides_2.copy().dropna()

nan_rows = taxi_rides_2_clean.loc[
    taxi_rides_2_clean["VendorID"].isna() | \
    taxi_rides_2_clean["trip_distance"].isna() |
    taxi_rides_2_clean["tip_amount"].isna() | \
    taxi_rides_2_clean["total_amount"].isna()
].shape[0]

print(f"Number of rows with NaN values: {nan_rows:,.0f}\n"
      f"Percentage of of rows with NaN values: {(nan_rows / taxi_rides_2_clean.shape[0]) * 100:.2f}%")
````

```VendorID``` must be ```np.int8```, ```trip_distance```, ```tip_amount```, and ```total_amount``` can be ```np.float32```

````python
taxi_rides_2_clean["VendorID"] = taxi_rides_2_clean["VendorID"].astype(np.int8)
taxi_rides_2_clean["trip_distance"] = taxi_rides_2_clean["trip_distance"].astype(np.float32)
taxi_rides_2_clean["tip_amount"] = taxi_rides_2_clean["tip_amount"].astype(np.float32)
taxi_rides_2_clean["total_amount"] = taxi_rides_2_clean["total_amount"].astype(np.float32)

taxi_rides_2_clean.head(10)
````

* Instead of removing ```NaN``` values from the ```VendorID``` column, set it to a new value: **3**. How does that affect your specifications and cleaning of the data?

````python
taxi_rides_2.loc[taxi_rides_2["VendorID"].isna(), "VendorID"] = 3

nan_rows = taxi_rides_2.loc[
    taxi_rides_2["VendorID"].isna() | \
    taxi_rides_2["trip_distance"].isna() |
    taxi_rides_2["tip_amount"].isna() | \
    taxi_rides_2["total_amount"].isna()
].shape[0]

print(f"Number of rows with NaN values: {nan_rows:,.0f}\n"
      f"Percentage of of rows with NaN values: {(nan_rows / taxi_rides_2.shape[0]) * 100:.2f}%")
````

Making this change there are not more ```NaN``` values in the dataset, and is not necessary to clean the data.
