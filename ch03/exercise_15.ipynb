# Exercise 15 - Weird taxi rides

For this exercise, I want you to create a data frame from the CSV data for January 2019:

1.  Load the CSV file into a data frame using only these four columns: ```passenger_count```, ```trip_distance```, ```payment_type```, and ```total_amount```.

````python
# NOTE: file nyc_taxi_2019-01.csv is not included due to its size. Get it from book's code repository

import pandas as pd

taxi_rides = pd.read_csv(
    "./nyc_taxi_2019-01.csv",
    header=0, # column names are on the first row of the file
    usecols=["passenger_count", "trip_distance", "payment_type", "total_amount"], # load only these columns
)

taxi_rides.head(10)
````

2. How many taxi rides had more than eight passengers?

````python
rides_8 = taxi_rides.loc[taxi_rides["passenger_count"] > 8].shape[0]

print(f"taxi rides with more than 8 passengers: {rides_8}")
````

3. How many taxi rides had zero passengers?

````python
rides_0 = taxi_rides.loc[taxi_rides["passenger_count"] == 0].shape[0]

print(f"taxi rides with 0 passengers: {rides_0}")
````

4. How many taxi rides were paid for in cash and cost over $1,000?

````python
CASH = 2
cash_over_1000 = taxi_rides.loc[(taxi_rides["payment_type"] == CASH) & (taxi_rides["total_amount"] > 1000)].shape[0]

print(f"taxi rides paid in cash and cost over $1000: {cash_over_1000}")
````

5. How many rides cost less than $0?

````python
cost_less_than_0 = taxi_rides.loc[taxi_rides["total_amount"] < 0].shape[0]

print(f"taxi rides that cost less than $0: {cost_less_than_0}")
````

6. How many rides traveled a below-average distance but cost an above-average amount?

````python
weird_rides = taxi_rides.loc[
    (taxi_rides["trip_distance"] < taxi_rides["trip_distance"].mean()) & \
    (taxi_rides["total_amount"] > taxi_rides["total_amount"].mean())
].shape[0]

print(f"taxi rides below average trip distance and above average cost: {weird_rides}")
````

## Beyond the exercise

* How many rides that cost less than $0 involved either a dispute (```payment_type``` of 4) or a voided trip (```payment_type``` of 6)?

````python
DISPUTE = 4
VOIDED_TRIP = 6

rides_disputed_or_voided = taxi_rides.query("total_amount < 0 & (payment_type == @DISPUTE | payment_type == @VOIDED_TRIP)").shape[0]

print(f"taxi rides disputed or voided: {rides_disputed_or_voided}")
````

* I stated earlier that most people pay for taxi rides using a credit card. Show this, and find what percentage normally pays in cash versus a credit card.

````python
CREDIT_CARD = 1
CASH = 2
rides_credit_card = taxi_rides.query("payment_type == @CREDIT_CARD").shape[0]
rides_credit_card_pct = (rides_credit_card / taxi_rides.shape[0]) * 100
rides_cash = taxi_rides.query("payment_type == @CASH").shape[0]
rides_cash_pct = (rides_cash / taxi_rides.shape[0]) * 100

print(f"taxi rides paid with credit card: {rides_credit_card}\n"
      f"percentage of rides paid with credit card: {rides_credit_card_pct:.2f}%\n"
      f"percentage of rides paid in cash: {rides_cash_pct:.2f}%\n"
      )
````
