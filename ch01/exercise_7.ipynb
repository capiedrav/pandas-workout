# Exercise 7 - Long, medium, and short taxi rides

In this exercise, we once again look at taxi data—but instead of the number of passengers, we examine the distance (in miles) each taxi ride went. Once again, I’ll ask youto create a series based on a single-column CSV file, **taxi-distance.csv**.

First, load the data into a series. Then modify the series (or create another series) containing category names rather than numbers based on these criteria:

* Short, <= 2 miles
* Medium, > 2 miles but <= 10 miles
* Long, > 10 miles

Calculate the number of rides in each category.

````python
import pandas as pd

# read file
taxi_ride_distances = pd.read_csv("./taxi-distance.csv", header=None).squeeze() # from DataFrame to Series

# create a new series for storing the ride categories

taxi_ride_categories = taxi_ride_distances.astype(str) # create a new series with the same length that taxi_ride_distances
taxi_ride_categories.loc[:] = "medium" # init all entries as medium
# short rides
taxi_ride_categories[taxi_ride_distances <= 2] = "short"
# long rides
taxi_ride_categories[taxi_ride_distances > 10] = "long"

print(f"First ten taxi rides distances (miles):\n{taxi_ride_distances.loc[:10]}\n\n"
      f"First ten taxi rides categories:\n{taxi_ride_categories.loc[:10]}")
````

* Calculate the number of rides in each category

````python
totals = taxi_ride_categories.value_counts()
print(f"Total of short rides: {totals.loc['short']}\n"
      f"Total of medium rides: {totals.loc['medium']}\n"
      f"Total of long rides: {totals.loc['long']}")
````

## Alternative solution

````python
totals = pd.cut(
    x=taxi_ride_distances,
    bins=[0, 2, 10, taxi_ride_distances.max()], # (0, 2] -> short; (2, 10] -> medium ; (10, max] -> long
    labels=["short", "medium", "long"],
    include_lowest=True,
    ).value_counts()

print(f"Total of short rides: {totals.loc['short']}\n"
      f"Total of medium rides: {totals.loc['medium']}\n"
      f"Total of long rides: {totals.loc['long']}")
````

## Beyond the exercise

* Compare the mean and median trip distances. What does that tell you about the distribution of the data?

````python
print(f"mean ride distance: {taxi_ride_distances.mean():.1f} miles\n"
      f"median ride distance: {taxi_ride_distances.median():.1f} miles\n")
````

**Answer:** The mean and median are far apart, which indicates that there are some long rides that skew the mean (outliers).
