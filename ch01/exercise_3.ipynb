# Exercise 3 - Counting tens digits

In this exercise, I want you to generate 10 random integers in the range 0 to 100. (Remember that the ```np.random.randint``` function returns numbers that include the lower bound but exclude the upper bound.) Create a series containing those numbers’ tens digits. Thus, if our series contains 10, 25, 32, we want the series 1, 2, 3.

````python
from pandas import Series
import numpy as np

# create random number generator
ng = np.random.default_rng(seed=0)

# create series
random_n = Series(ng.integers(0, 100, size=10, dtype=np.int8))

tens = random_n // 10 # obtain tens
print(f"random numbers:{random_n.values}\ntens:{tens.values}")
````

## Beyond the exercise

*  What if the range were from 0 to 10,000? How would that change your strategy,
if at all?

````python
new_random = Series(ng.integers(0,10000, size=10, dtype=np.int16))

# to obtain the tens, first obtain the hundreds
hundreds_and_tens = new_random % 100
# and from them, obtain the tens
tens = hundreds_and_tens // 10

print(f"random numbers and their tens:")
for i in new_random.index:
    print(f"{new_random[i]} - {tens[i]}")
````

* Create a new series with 10 floating-point values between 0 and 1,000. Find the numbers whose integer component (i.e., ignoring any fractional part) are even.

````python
floats = Series(ng.random((10, ), dtype=np.float32) * 1000) # ten floats between 0 and 1000
print(f"ten floats between 0 and 1000:\n{floats}\n")

mask_int_even = np.floor(floats) % 2 == 0 # find numbers with int component even (mask index)

print(f"numbers with int component even:\n{floats[mask_int_even]}")
````
