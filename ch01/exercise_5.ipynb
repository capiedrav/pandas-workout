# Exercise 5 - Monday temperatures

In this exercise, I want you to create a series of 28 temperature readings in Celsius, representing four seven-day weeks, randomly selected from a normal distribution with a mean of 20 and a standard deviation of 5, rounded to the nearest integer. The index should
start with Sun, continue through Sat, and repeat Sun through Sat until each temperature has a value. The question is, **what was the mean temperature on Mondays during this period?**


```python
from pandas import Series
import numpy as np

# create temperature readings
ng = np.random.default_rng(seed=0)
temperatures = Series(np.rint((ng.normal(20, 5, 28))), index=("Sun Mon Tue Wed Thu Fri Sat " * 4).split(), dtype=np.int8)

print(temperatures)
```

* what was the mean temperature on Mondays during this period?

```python
mean_temperature = temperatures.loc["Mon"].mean()
print(f"Monday's mean temperature: {mean_temperature:.1f} °C")
```

## Beyond the exercise

* What was the average weekend temperature (i.e., Saturdays and Sundays)?

```python
mean_weekend_temperature = temperatures.loc[["Sat", "Sun"]].mean()

print(f"Mean weekend temperature: {mean_weekend_temperature:.1f} °C")
```

* How many times is the change in temperature from the previous day greater than 2 degrees?

```python
n = 0
for i in range(1, len(temperatures)):
    delta_t = np.abs(temperatures.iloc[i] - temperatures.iloc[i - 1]) # change in temperature from the previous day
    if delta_t > 2:
        n += 1

print(f"Times the change in temperature from previous day is greater than 2 °C: {n}")
```

* What are the two most common temperatures in our data set, and how often does each appear?

```python
print(f"Two most common temperatures:\n{temperatures.value_counts(sort=True, ascending=False)[:2]}\n")
```
