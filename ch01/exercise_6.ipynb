# Exercise 6 - Passenger Frequency

In this exercise, we begin looking at real-world data loaded from a one-column CSV file. We’ll take a deeper look at reading from and writing to files in chapter 3, but we start here by invoking ```pd.read_csv```, calling ```squeeze``` on the one-column data frame it returns, and getting back a series.

The data is in the file **taxi-passenger-count.csv**, available along with the other data files used in this course. The data comes from 2015 data I retrieved from New York City’s open data site, where you can get enormous amounts of information about taxi rides in New York City over the last few years. This file shows the number of passengers in each of 100,000 rides.

Your task in this exercise is to show what percentage of taxi rides had only one passenger versus the (theoretical) maximum of six passengers.

````python
import pandas as pd
import numpy as np

taxi_passengers = pd.read_csv("./taxi-passenger-count.csv", header=None, dtype=np.int8).squeeze() # from Dataframe to Series

print(f"First ten rides:\n{taxi_passengers.iloc[:10]}\n")

one_passenger_rides_pct = (taxi_passengers[taxi_passengers == 1].count() / taxi_passengers.count()) * 100
six_passenger_rides_pct = (taxi_passengers[taxi_passengers == 6].count() / taxi_passengers.count()) * 100

print(f"percentage of one passenger rides: {one_passenger_rides_pct:.1f}%\n"
      f"percentage of six passenger rides: {six_passenger_rides_pct:.1f}%")
````

## Alternative solution

````python
one_and_six_passenger_rides_pct = taxi_passengers.value_counts(normalize=True)[[1, 6]] * 100

print(f"percentage of one passenger rides: {one_and_six_passenger_rides_pct.iloc[0]:.1f}%\n"
      f"percentage of six passenger rides: {one_and_six_passenger_rides_pct.iloc[1]:.1f}%")
````

## Beyond the exercise

*  What are the 25%, 50% (median), and 75% quantiles for this data set? Can you guess the results before you execute the code?

````python
print(f"25% quantile: {taxi_passengers.quantile(0.25)}\n"
      f"50% quantile (median): {taxi_passengers.median()}\n"
      f"75% quantile: {taxi_passengers.quantile(0.75)}")
````

* What proportion of taxi rides are for three, four, five, or six passengers?

````python
one_or_two_passenger_rides_pct = (taxi_passengers.value_counts(normalize=True)[[1, 2]] * 100).sum()
print(f"percentage of three, four, five or six passenger rides: {(100 - one_or_two_passenger_rides_pct):.1f}%")
````
