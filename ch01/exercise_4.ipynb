# Exercise 4 - Descriptive Statistics

The mean, median, and standard deviation are three numbers we can use to get a better picture of our data. Adding a few other numbers can give us an even more complete picture. These _descriptive statistics_ typically include the mean, standard deviation, minimum value, 25% quantile, median, 50% quantile, and maximum value. Understanding and using descriptive statistics is a key skill for anyone working with data, and in this exercise, you’ll practice doing so with the following:

* Generate a series of 100,000 floats in a normal distribution with a mean of 0 and a standard deviation of 100.

```python
from pandas import Series
import numpy as np

ng = np.random.default_rng(seed=0)
mu = 0 # mean
std = 100 # standard deviation
size = 100_000 # sample size
sample = Series(ng.normal(mu, std, size), dtype='float64')
```

* Get the descriptive statistics for this series. How close are the mean and median? (You don’t need to calculate the difference; rather, consider why they aren’t the same.)

```python
print(f"descriptive statistics:\n{sample.describe()}\n") # descriptive statistics
print(f"sample mean: {sample.mean()}\nsample median: {sample.median()}\nsample std: {sample.std()}")
```

* Replace the minimum value with 5 times the maximum value.

```python
# use a boolean mask to find all occurrences of sample's minimum value
sample.loc[sample == sample.min()] = 5 * sample.max() # replace each occurrence with 5 times sample's max
```

*  Get the descriptive statistics again. Did the mean, median, and standard deviations change from their previous values? (Again, it’s enough to see the difference without calculating it.) If so, why?

```python
print(f"sample mean: {sample.mean()}\nsample median: {sample.median()}\nsample std: {sample.std()}")
```

## Beyond the exercise

* Demonstrate that 68%, 95%, and 99.7% of the values in s are indeed within one, two, and three standard distributions of the mean.

```python
# reset the sample
sample = Series(ng.normal(mu, std, size), dtype='float64')

percentage_values_in_one_std = len(sample.loc[(-(sample.std() + sample.mean()) < sample) & (sample < sample.mean() + sample.std())]) / len(sample)

percentage_values_in_two_std = len(sample.loc[(-2 * (sample.std() + sample.mean()) < sample) & (sample < 2 * (sample.mean() + sample.std()))]) / len(sample)

percentage_values_in_three_std = len(sample.loc[(-3 * (sample.std() + sample.mean()) < sample) & (sample < 3 * (sample.mean() + sample.std()))]) / len(sample)

print(f"percentage values in one standard deviation: {percentage_values_in_one_std * 100:.2f} %")
print(f"percentage values in two standard deviations: {percentage_values_in_two_std * 100:.2f} %")
print(f"percentage values in three standard deviations: {percentage_values_in_three_std * 100:.2f} %")
```

*  Calculate the mean of numbers greater than s.mean(). Then calculate the mean of numbers less than s.mean(). Is the average of these two numbers the same as s.mean()?

```python
mean_values_gt_mean = sample.loc[sample > sample.mean()].mean()
mean_values_lt_mean = sample.loc[sample < sample.mean()].mean()
avg = (mean_values_gt_mean + mean_values_lt_mean) / 2

print(f"mean of values greater than mean: {mean_values_gt_mean}")
print(f"mean of values less than mean: {mean_values_lt_mean}")
print(f"average of both values: {avg}")
```

Is the average of these two numbers the same as s.mean()?

**Answer**: yes.
