# Exercise 23 - Olympic games

In this exercise, we’re going to build a deep multi-index, allowing us to retrieve data from various levels and in several ways. Specifically, I want you to do the following:

1.  Read the data file (```olympic_athlete_events.csv```) into a data frame. We only care about some of the columns: ```Age```, ```Height```, ```Team```, ```Year```, ```Season```, ```City```, ```Sport```, ```Event```, and ```Medal```. The multi-index should be based on ```Year```, ```Season```, ```Sport```, and ```Event```.

````python
import pandas as pd

olympic_data = pd.read_csv(
    filepath_or_buffer="./olympic_athlete_events.csv", # NOTE: make sure this file is in this folder
    header=0,
    usecols=["Age", "Height", "Team", "Year", "Season", "City", "Sport", "Event", "Medal"],
    index_col=["Year", "Season", "Sport", "Event"],
)

olympic_data.head(10)
````

2. Answer these questions:

- What is the average age of winning athletes in summer games held between 1936 and 2000?

````python
# let's sort the indexes
olympic_data = olympic_data.sort_index()

olympic_data.head(10)
````

````python
avg_age_winners = olympic_data.loc[(slice(1936, 2000), "Summer"), ["Age", "Medal"]].query('Medal == "Gold"')["Age"].mean()

print(f"Average Age of winners in summers games 1936-2000: {avg_age_winners:.1f} years old")
````

- What team has won the most medals in all archery events?

````python
team_most_medals = olympic_data.loc[
    (
        slice(None), # include all years
        slice(None), # include all seasons
        "Archery"
    ),
    ["Team", "Medal"]
].query('Medal == "Gold"')["Team"].value_counts().head(1)

team_most_medals
````

* Starting in 1980, what is the average height of the “Table Tennis Women’s Team” event?

````python
avg_height = olympic_data.loc[
    (
        slice(1980, None), # 1980 onwards
        slice(None), # include all seasons
        "Table Tennis",
        "Table Tennis Women's Doubles"
    ),
    "Height"
].mean()

print(f"Average height of the Table Tennis Women's Doubles: {avg_height:.1f} cm")
````

* Starting in 1980, what is the average height of both “Table Tennis Women’s Team” and “Table Tennis Men’s Team”?

````python
avg_height = olympic_data.loc[
    (
        slice(1980, None), # 1980 onwards
        slice(None), # include all seasons
        "Table Tennis",
        ["Table Tennis Women's Doubles", "Table Tennis Men's Doubles"]
    ),
    "Height"
].mean()

print(f"Average height of the Table Tennis Women/Men Doubles: {avg_height:.1f} cm")
````

* How tall was the tallest-ever tennis player in Olympic games from 1980 until 2016?

````python
height_tallest_tennis_player = olympic_data.loc[
    (
        slice(1980, 2016), # from 1980 to 2016
        slice(None), # include all seasons
        "Tennis",
        "Tennis Men's Singles"
    ),
    "Height"
].max()

print(f"Height of tallest tennis player from 1980-2016: {height_tallest_tennis_player:.1f} cm")
````

## Beyond the exercise

*  Events occur in either summer or winter Olympic Games, but not both. As a result, the "```Season```" level in our multi-index is often unnecessary. Remove the "```Season```" level, and then find (again) the height of the tallest tennis player between 1980 and 2016.

````python
# remove "Season" from multi-index (and from dataframe)
olympic_data = olympic_data.droplevel(level="Season")

olympic_data.head(10)
````

````python
# find answer using xs method
height_tallest_tennis_player = olympic_data.xs(
    key=(slice(1980, 2016), "Tennis Men's Singles"),
    level=["Year", "Event"]
)["Height"].max()

print(f"Height of tallest tennis player from 1980-2016: {height_tallest_tennis_player:.1f} cm")
````

* In which city were the most gold medals awarded from 1980 onward?

````python
# using IndexSlice
from pandas import IndexSlice as idx
city_gold = olympic_data.loc[idx[
    1980:, # from 1980 onwards
    :, # include all sports
    :, # include all events
    ],
    ["City", "Medal"] # select this columns
].query("Medal == 'Gold'")["City"].value_counts().head(1) # select city with more gold medals

city_gold
````

* How many gold medals were received by the United States since 1980? (Use the index to select the values.)

````python
usa_golds = olympic_data.loc[
    idx[
        1980:, # from 1980 onwards
        :, # include all sports
        :, # include all events
    ],
    ["Team", "Medal"] # select this columns
].query("(Team == 'United States') & (Medal == 'Gold')").shape[0]

print(f"USA Gold medals since 1980: {usa_golds:,.0f}")
````
