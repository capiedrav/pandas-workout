# Exercise 21 - Parking tickets

In this exercise, I’ll ask you to perform some queries on another data set from New York City: one that tracked all parking tickets during the year 2020—more than 12 million of them. You could, in theory, perform these queries without modifying the data frame’s index. However, I want you to get some practice setting and resetting the index. We’re going to do that a lot in this chapter, and you’ll likely do it a great deal as you work with pandas with real-life data sets.

With that in mind, I want you to do the following:
1. Create a data frame from the file ```nyc-parking-violations-2020.csv```. We are only interested in a handful of the columns:
* ```Date First Observed```
* ```Plate ID```
* ```Registration State```
* ```Issue Date``` (a string in MM/DD/YYYY format, always followed by ```12:00:00AM```)
* ```Vehicle Make```
* ```Street Name```
* ```Vehicle Color```

````python
import pandas as pd


nyc_data = pd.read_csv(
    filepath_or_buffer='./nyc-parking-violations-2020.csv', # NOTE: make sure this file is in the current folder
    usecols=[
        "Date First Observed",
        "Plate ID",
        "Registration State",
        "Issue Date",
        "Vehicle Make",
        "Street Name",
        "Vehicle Color"
    ]
)

nyc_data.head(10)
````

2. Set the data frame’s index to the ```Issue Date``` column.

````python
nyc_data = nyc_data.set_index("Issue Date")

nyc_data.head(10)
````

3. Determine what three makes were most frequently ticketed on January 2, 2020.

````python
most_ticketed_makes = nyc_data.loc["01/02/2020 12:00:00 AM", "Vehicle Make"].value_counts().head(3)

most_ticketed_makes
````

4. Determine the five streets on which cars got the most tickets on June 1, 2020.

````python
most_ticketed_streets = nyc_data.loc["06/01/2020 12:00:00 AM", "Street Name"].value_counts().head(5)

most_ticketed_streets
````

5. Set the index to ```Vehicle Color```.

````python
nyc_data = nyc_data.reset_index().set_index("Vehicle Color")

nyc_data.head(10)
````

6. Determine the most common make of vehicles that were either red or blue.

````python
most_common_make_BLUE_RED = nyc_data.loc[["BLUE", "RED"], "Vehicle Make"].value_counts().head(1)

most_common_make_BLUE_RED
````
