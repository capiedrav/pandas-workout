# Exercise 24 - Olympic pivots

In this exercise, we examine the Olympic data one more time—but using pivot tables, so we can examine and compare more information at a time than we could before. Pivot tables are a popular way to summarize information in a larger, more complex
table. I want you to do the following:

1. Read in our Olympic data again.
* Only use these columns: ```Age```, ```Height```, ```Team```, ```Year```, ```Season```, ```Sport```, ```Medal```.
* Only include games from 1980 to the present.
* Only include data from these countries: Great Britain, France, United States, Switzerland, China, and India.

````python
import pandas as pd

olympic_data = pd.read_csv(
    filepath_or_buffer="./olympic_athlete_events.csv",
    header=0,
    usecols=["Age", "Height", "Team", "Year", "Season", "Sport", "Medal"]
)

olympic_data.head(10)
````

````python
# select rows of the following countries: Great Britain, France, United States, Switzerland, China, and India
olympic_data = olympic_data.loc[
    olympic_data["Team"].isin([
        "Great Britain",
        "France",
        "United States",
        "Switzerland",
        "China",
        "India"
    ])
]

# remove rows which "Year" is before 1980
olympic_data = olympic_data.loc[olympic_data["Year"] >= 1980]

olympic_data.head(10)
````

2. Answer the following questions:
*  What was the average age of Olympic athletes? In which country do players appear to consistently be the youngest?

````python
pd.options.display.float_format = '{:,.1f}'.format
olympic_data.pivot_table(index="Year", columns="Team", values="Age")
````

**Answer:** from the pivot table above, it seems China has the youngest athletes.

* How tall were the tallest athletes in each sport in each year?

````python
olympic_data.pivot_table(index="Sport", columns="Year", values="Height", aggfunc="max")
````

* How many medals did each country earn each year?

````python
# olympic_data = olympic_data.dropna(subset="Medal")

olympic_data.pivot_table(
    index="Year",
    columns="Team",
    values="Medal",
    aggfunc="count"
)
````

## Beyond the exercise

* Create a pivot table that shows the number of medals each team won per year, with the index including the year and the season in which the games took place.

````python
olympic_data.pivot_table(index=["Year", "Season"], columns="Team", values="Medal", aggfunc="count")
````

* Create a pivot table that shows both the average age and the average height per year per team.

````python
olympic_data.pivot_table(index="Year", columns="Team", values=["Age", "Height"])
````

* Create a pivot table that shows the average age and the average height per year, per team, broken up by year and season.

````python
olympic_data.pivot_table(
    index=["Year", "Season"],
    columns="Team",
    values=["Age", "Height"]
)
````
