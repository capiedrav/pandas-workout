# Exercise 22 - State SAT scores

In this exercise, we look at a summary of scores from the SAT, a standardized university admissions test widely used in the United States. The CSV file (```sat-scores.csv```) has 99 columns and 577 rows describing all 50 US states and three nonstates (Puerto Rico, the Virgin Islands, and Washington, DC) from 2005 through 2015. In this exercise, I want you to:

1. Read in the scores file, only keeping the ```Year```, ```State.Code```, ```Total.Math```, ```Total.Test-takers```, and ```Total.Verbal``` columns.

````python
from heapq import heappush

import pandas as pd

test_scores = pd.read_csv(
    filepath_or_buffer="./sat-scores.csv",
    header=0,
    usecols=["Year", "State.Code", "Total.Math", "Total.Test-takers", "Total.Verbal"]
)

test_scores.head(10)
````

2. Create a multi-index based on the year and the two-letter state code.

````python
test_scores = test_scores.set_index(["Year", "State.Code"])

test_scores.head(10)
````

3. Determine how many people took the SAT in 2005.

````python
n_test_takers_2005 = test_scores.loc[2005, "Total.Test-takers"].sum()

print(f"Test takers in 2005: {n_test_takers_2005:,.0f}")
````

4.  Determine the average SAT math score in 2010 from New York (NY), New Jersey (NJ), Massachusetts (MA), and Illinois (IL).

````python
avg_test_scores = test_scores.loc[([2010, ], ["NY", "NJ", "MA", "IL"]), "Total.Math"].mean()

print(f"Average math score in 2010 from NY, NJ, MA, IL: {avg_test_scores:,.2f}")
````

5. Determine the average SAT verbal score in 2012–2015 from Arizona (AZ), California ( CA), and Texas (TX).

````python
avg_test_scores = test_scores.loc[(slice(2012,2015), ["AZ", "CA", "TX"]), "Total.Verbal"].mean()

print(f"Average verbal score in 2012-2015 from AZ, CA, TX: {avg_test_scores:,.2f}")
````

## Beyond the exercise

````python
avg_test_scores = test_scores.loc[(slice(None), ["FL", "IN", "ID"]), ["Total.Math", "Total.Verbal"]].mean()

avg_test_scores
````

* Which state received the highest verbal score, and in which year?

````python
index = test_scores["Total.Verbal"].idxmax()

print(f"The state with the highest verbal score is: {index[1]}\n"
      f"Obtained in the year {index[0]}\n"
      f"with a score of {test_scores.loc[index, 'Total.Verbal']:,.2f}"
      )
````

* Was the average math score in 2005 higher or lower than that in 2015?

````python
math_score_2005 = test_scores.loc[2005, "Total.Math"].mean()
math_score_2015 = test_scores.loc[2015, "Total.Math"].mean()

print(f"Math score in 2005: {math_score_2005:,.2f}\n"
      f"Math score in 2015: {math_score_2015:,.2f}\n"
      f"The year with the higher math score is: {2015 if math_score_2015 > math_score_2005 else 2005}\n"
      )
````
