# Exercise 14 - Selective updating

In this exercise, I want you to create the same two-column data frame as in the last exercise. Then, update the values in the ```temperature``` column so that any value less than 0 is set to 0.

````python
import numpy as np
import pandas as pd

temperature = pd.read_csv('./nyc-temps.txt', dtype=np.float32).squeeze()
temp_df = pd.DataFrame(
  {
      "hour": [0, 3, 6, 9, 12, 15, 18, 21] * 91,
      "temperature": temperature
  })

temp_df.head(8)
````

* Update the values in the ```temperature``` column so that any value less than 0 is set to 0.

````python
temp_df_copy = temp_df.copy()
temp_df_copy.loc[temp_df_copy["temperature"] < 0, "temperature"]  = 0

temp_df_copy.head(8)
````

## Beyond the exercise

* Set all the odd temperatures to the mean of all the temperatures.

````python
temp_df_copy = temp_df.copy()
````

````python
temp_df_copy.loc[
    temp_df_copy["temperature"] % 2 == 1, # select odd temperatures
    "temperature"
] = temp_df_copy["temperature"].mean()

temp_df_copy.head(16)
````

*  Set the even temperatures at hours 9 and 18 to 3.

````python
temp_df_copy = temp_df.copy()
````

````python
temp_df_copy.loc[
    (temp_df_copy["temperature"] % 2 == 0) & (temp_df_copy["hour"].isin([9, 18])), # even temperatures at hours 9 and 18
    "temperature"
] = 3

temp_df_copy.head(16)
````

* If the hour is odd, set the temperature to 5.

````python
temp_df_copy = temp_df.copy()
````

````python
temp_df_copy.loc[temp_df_copy["hour"] % 2 == 1, "temperature"] = 5

temp_df_copy.head(16)
````
