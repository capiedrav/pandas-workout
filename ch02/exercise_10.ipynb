# Exercise 10 - Adding new products

Good news! Your store is making money, and you have decided to add some new products. I’d like you to do that by creating a new data frame and adding it to the existing one. This new data frame should contain three products (including product ID, name, wholesale price, and retail price):

* Phone, with an ID of 24, a wholesale price of 200, and a retail price of 500
* Desktop, with an ID of 16, a wholesale price of 1000, and a retail price of 1600
* Headphones, with an ID of 17, a wholesale price of 75, and a retail price of 110

Because these are new products, don’t include the ```sales``` column. And to avoid problems and conflicts, ensure that the indexes of these new products are different from existing product indexes. (In chapter 4, we’ll look at some ways to handle index problems more elegantly.)

Once you have added these new products, assign sales figures to each of them. Finally, recalculate the store’s total net income, including the new products.

1. Recreate inventory from exercise 9

````python
import pandas as pd
from pandas import DataFrame

pd.options.display.float_format = "{:,.2f}".format # use commas for thousands separator, and 2 decimal places

inventory = DataFrame(
    [
      [11, "keyboard", 10.00, 15.00, 20],
      [55, "Mouse", 5.00, 10.00, 32],
      [30, "Bluetooth Speaker", 20.00, 35.00, 50],
      [25, "Tablet", 200.00, 450.00, 10],
      [13, "Laptop", 500.00, 800.00, 36],
     ],
    columns=["product_id", "product_name", "wholesale_price", "retail_price", "sales"],
)

print(inventory)
````

2. Add new products to inventory

````python
new_products = DataFrame(
    [
        [24, "Phone", 200.00, 500.00],
        [16, "Desktop", 1000.00, 1700.00],
        [17, "Headphones", 75.00, 110.00]
    ],
    columns=["product_id", "product_name", "wholesale_price", "retail_price"],
    index=range(5, 8) # avoid collisions with the index of inventory DataFrame
)

inventory = pd.concat([inventory, new_products])

inventory
````

3. Fill in sales info for the new products

````python
inventory.loc[[5, 6, 7], "sales"] = [10, 20, 50]

inventory
````

4. Calculate revenue of each product

````python
inventory["revenue"] = (inventory["retail_price"] - inventory["wholesale_price"]) * inventory["sales"]


print(f"inventory:\n{inventory}\n\nTotal revenue: ${inventory['revenue'].sum():,.2f}")
````
