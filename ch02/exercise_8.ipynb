# Exercise 8 - Net revenue

For this exercise, I want you to create a data frame that represents a company’s inventory of five products. Each product has a unique ID number (a two-digit integer will do), name, wholesale price, retail price, and number of sales in the last month. You’re making it up, so if you’ve always wanted to be a profitable starship dealer, this is your chance! **Once you have created this data frame, calculate the total net revenue from all your products**.

```python
from pandas import DataFrame

inventory = DataFrame(
    [
      [11, "keyboard", 10.00, 15.00, 20],
      [55, "Mouse", 5.00, 10.00, 32],
      [30, "Bluetooth Speaker", 20.00, 35.00, 50],
      [25, "Tablet", 200.00, 450.00, 10],
      [13, "Laptop", 500.00, 800.00, 36],
     ],
    columns=["product_id", "product_name", "wholesale_price", "retail_price", "sales"]
)

print(inventory)
```

* Calculate the total net revenue from all your products.

```python
revenue = ((inventory["retail_price"] - inventory["wholesale_price"]) * inventory["sales"]).sum()

print(f"Net revenue from all products: ${revenue:.1f}")
```

# Beyond the exercise

* For what products is the retail price more than twice the wholesale price?

```python
print(f"Answer:\n{inventory[inventory['retail_price'] > 2 * inventory['wholesale_price']][['product_id', 'product_name']]}")
```

*  How much did the store make from computer accessories versus tablets and laptops? (You can retrieve based on the index values, not anything more sophisticated.)

```python
revenue_accessories = ((inventory["retail_price"].iloc[:3] - inventory["wholesale_price"].iloc[:3]) * inventory["sales"].iloc[:3]).sum()
revenue_tablets_and_laptops = revenue - revenue_accessories

print(f"Revenue accessories: ${revenue_accessories}\nRevenue tablets and laptops: ${revenue_tablets_and_laptops}\n"
      f"Revenue accessories (%):{(revenue_accessories / revenue) * 100:.1f}%")
```

* Because your store is doing so well, you can negotiate a 30% discount on the wholesale price of goods. Calculate the new net income.

```python
inventory["wholesale_price"] = inventory["wholesale_price"] * 0.7 # discount of 30% on the wholesale price of all products

new_revenue = ((inventory["retail_price"] - inventory["wholesale_price"]) * inventory["sales"]).sum()

print(f"Net revenue from all products: ${new_revenue:.1f}")
```
