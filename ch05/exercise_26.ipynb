# Exercise 26 - Celebrity deaths

For this exercise, we’ll look at a (slightly morbid) data set: a list of celebrities who died in 2016 and whose passing was recorded in Wikipedia, including the date of death, a short biography, and the cause of death. The problem is that this data set is messy, with some missing data and some erroneous data that will prevent us from working with it easily. The goal of this exercise is to find the average age of celebrities who died February–July 2016. Getting there will take several steps:

1. Create a data frame from the file ```celebrity_deaths_2016.csv```. For this exercise, we’ll use only two columns: ```ageofdeath``` and ```age```.

````python
import pandas as pd

dead_celebrities = pd.read_csv(
    "celebrity_deaths_2016.csv",
    header=0,
    usecols=["dateofdeath", "age"]
)

dead_celebrities.head(10)
````

2. Create a new ```month``` column containing the month from the ```dateofdeath``` column.

````python
dead_celebrities["month"] = dead_celebrities["dateofdeath"].str.slice(5, 7)
dead_celebrities.head(10)
````

3. Make the ```month``` column the index of the data frame.

````python
dead_celebrities = dead_celebrities.set_index("month")
dead_celebrities.head(10)
````

4. Sort the data frame by the index.

````python
dead_celebrities = dead_celebrities.sort_index()
dead_celebrities.head(10)
````

5. Clean all nonintegers from the age column.

````python
dead_celebrities = dead_celebrities.loc[dead_celebrities["age"].str.isdigit()]
dead_celebrities.head(10)
````

6. Turn the ```age``` column into an integer value.

````python
import numpy as np

dead_celebrities["age"] = dead_celebrities["age"].astype(np.int64)
dead_celebrities.head(10)
````

7. Find the average age of celebrities who died during that period (Feb-Jul 2016).

````python
mean_age = dead_celebrities.loc["02":"07", "age"].mean()

print(f"Mean age of dead celebrities between Feb and Jul of 2016: {mean_age:.1f} years old.")
````

## Beyond the exercise

*  Add a new column, ```day```, from the day of the month in which the celebrity died. Then create a multi-index (from ```month``` and ```day```). What was the average age of death from Feb. 15 through July 15?

````python
# create day column
dead_celebrities["day"] = dead_celebrities["dateofdeath"].str.slice(8, 10)
````

````python
# create multi-index
dead_celebrities = dead_celebrities.reset_index()
dead_celebrities = dead_celebrities.set_index(["month", "day"])
dead_celebrities = dead_celebrities.sort_index()
dead_celebrities.head(10)
````

````python
mean_age = dead_celebrities.loc[("02", "15"):("07", "15"), "age"].mean()

print(f"Mean age of celebrity deaths between Feb 15 and Jul 15 of 2016: {mean_age:.1f} years old.")
````
